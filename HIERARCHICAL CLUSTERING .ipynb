```python



# Step 1: Import required libraries
import matplotlib.pyplot as plt
import csv
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

print("Libraries imported successfully!")
```

```python

# Step 2: Load the Iris dataset
file_path = r"C:\Users\KRUTIKA SHIVANAND\Downloads\Iris.csv"

with open(file_path, "r") as file:
    reader = csv.reader(file)
    header = next(reader)

    data = []
    for row in reader:
        data.append([float(row[1]), float(row[2])])

X = np.array(data)

print("\nDataset Shape:", X.shape)
print("\nFirst 5 rows:")
print(X[:5])
```

```python


# Step 3: Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\nData standardized successfully!")
```

```python

# Step 4: Create a dendrogram
linked = linkage(X_scaled, method='ward')

plt.figure(figsize=(12, 6))
dendrogram(linked)

plt.title("Dendrogram - Hierarchical Clustering")
plt.xlabel("Data Points")
plt.ylabel("Euclidean Distance")
plt.grid(axis='y')
plt.show()
```

```python
# Step 5: Apply Agglomerative Clustering
model = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

clusters = model.fit_predict(X_scaled)

print("\nAgglomerative Clustering completed!")
```

```python
# Step 6: Visualize the clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    cmap='viridis',
    s=50
)

plt.xlabel("Standardized Sepal Length")
plt.ylabel("Standardized Sepal Width")
plt.title("Agglomerative Hierarchical Clustering")
plt.grid(True)
plt.show()

# Step 7: Display cluster assignments
print("\nCluster Assignments:")
print(clusters)

print("\nFirst 10 Data Points and Their Clusters:")

for i in range(10):
    print("Data Point:", i, "Cluster:", clusters[i])
```
